## Data Cleaning and Preprocessing

### Cleaning `df_assessments`

```python
print('Displaying info for df_assessments before cleaning:')
df_assessments.info()
print('Displaying missing values in df_assessments before cleaning:')
display(df_assessments.isnull().sum())

# Replacing '?' with pandas.NA in 'date' column
# This line is safer as it directly replaces without problematic check on unique() if pd.NA is present
df_assessments['date'] = df_assessments['date'].replace('?', pd.NA)

# Converting 'date' column to numeric (days from start of module), coercing errors to NaN
df_assessments['date'] = pd.to_numeric(df_assessments['date'], errors='coerce')

# 'weight' column is already numeric

print('Checking for duplicates in df_assessments:')
print(f"Number of duplicate rows: {df_assessments.duplicated().sum()}")
if df_assessments.duplicated().any():
    # Dropping duplicate rows.
    df_assessments.drop_duplicates(inplace=True)
    print('Duplicate rows removed.')

print('Displaying info for df_assessments after cleaning:')
df_assessments.info()
display(df_assessments.head())
```

## Data Merging

First, `df_studentInfo` and `df_studentRegistration` are merged. These two tables share `code_module`, `code_presentation`, and `id_student`, which together uniquely identify a student's enrollment in a specific course presentation. Using a `left`, the merge will keep all students from `df_studentInfo`.

```python
# Merging df_studentInfo and df_studentRegistration on common keys
df_student_details = pd.merge(
    df_studentInfo,
    df_studentRegistration,
    on=['code_module', 'code_presentation', 'id_student'],
    how='left'
)
display(df_student_details.head())
```

Next, the assessment details from `df_assessments` are linked to the student assessment records in `df_studentAssessments` with merge on `id_assessment`.. This will bring in information on `assessment_type`, `date` (relative to course start), and `weight` for each student's assessment submission.

```python
# Merging df_studentAssessments and df_assessments on 'id_assessment'
df_student_assessment_details = pd.merge(
    df_studentAssessments,
    df_assessments,
    on='id_assessment',
    how='left'
)
display(df_student_assessment_details.head())
```

Finally, the `df_student_details` (containing student info and registration) is merged with `df_student_assessment_details` (containing student assessment records with course context). The common keys for this merge are `code_module`, `code_presentation`, and `id_student`.

```python
# Merging df_student_details and df_student_assessment_details on common keys
df_merged = pd.merge(
    df_student_details,
    df_student_assessment_details,
    on=['code_module', 'code_presentation', 'id_student'],
    how='left'
)

print("Displaying merged DataFrame Info:")
df_merged.info()
print("\nDisplaying first 5 rows of the merged DataFrame:")
display(df_merged.head())
```

Since directly merging the large `df_studentVle` DataFrame caused a memory issue, the data needs to be preprocessed by aggregation. The data, `df_studentVle`, is grouped by student, course presentation, module, and VLE item (`id_site`) and sum their `sum_click` interactions. This reduced the size of the DataFrame before merging.

```python
# Aggregating df_studentVle to reduce its size before merging
df_studentVle_agg = df_studentVle.groupby(['code_module', 'code_presentation', 'id_student', 'id_site'])['sum_click'].sum().reset_index()

print("Displaying aggregated df_studentVle Info:")
df_studentVle_agg.info()
print("\nDisplaying first 5 rows of the aggregated df_studentVle:")
display(df_studentVle_agg.head())
```

```python
# Merging df_merged with aggregated df_studentVle_agg using a left merge
df_merged = pd.merge(
    df_merged,
    df_studentVle_agg,
    on=['code_module', 'code_presentation', 'id_student'],
    how='left'  # Using a left merge to keep all existing student records
)

print("Displaying merged DataFrame Info after adding aggregated df_studentVle:")
df_merged.info()
print("\nDisplaying first 5 rows of the merged DataFrame after adding aggregated df_studentVle:")
display(df_merged.head())
```

```python
# Saving df_merged dataset to CSV
df_merged.to_csv('df_merged.csv', index=False)
```

### EDA 4.2: Smoothed Trend of Average Assessment Score by Final Result Over Time (Highlighting 'Withdrawn')

This plot applies a rolling mean to the average assessment scores for each final result group. This smoothing helps to reduce noise and reveal underlying trends more clearly. Additionally, the 'Withdrawn' category is given a prominent color to emphasize its trajectory, making it easier to identify this highly at-risk group.

```python
# Recalculating average scores by date and final result
avg_scores_by_date = df_merged.groupby(['date', 'final_result'])['score'].mean().reset_index()

# Sorting by date for correct rolling mean calculation
avg_scores_by_date = avg_scores_by_date.sort_values(by='date')

# Applying a rolling mean within each final_result group for smoothing
# setting a window size of 10 for smoothing
avg_scores_by_date['smoothed_score'] = avg_scores_by_date.groupby('final_result')['score'].transform(lambda x: x.rolling(window=10, min_periods=1).mean())

# Defining a custom palette to highlight 'Withdrawn'
custom_palette = {
    'Withdrawn': 'red',
    'Fail': 'orange',
    'Pass': 'green',
    'Distinction': 'blue'
}

plt.figure(figsize=(14, 8))
sns.lineplot(
    x='date', y='smoothed_score', hue='final_result', data=avg_scores_by_date.dropna(subset=['smoothed_score']),
    palette=custom_palette, marker='o', linewidth=2.5
)
plt.title('Smoothed Average Assessment Score Trend by Final Result Over Time (Withdrawn Highlighted)')
plt.xlabel('Assessment Date (Days from Module Start)')
plt.ylabel('Smoothed Average Score')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Final Result')
plt.tight_layout()
plt.show()
```

### Insight from EDA 4.2 (Smoothed & Highlighted):

*   **DESCRIBE:** The line plot shows the smoothed average assessment `score` trend over `date` for all `final_result` categories, with the 'Withdrawn' trend line distinctly highlighted in red. A rolling mean has been applied to smooth out short-term fluctuations.
*   **INTERPRET:** The smoothed trends provide a much clearer and more reliable view of the long-term performance trajectories. The prominent red 'Withdrawn' line vividly illustrates that these students frequently maintain a lower average score throughout the module or experience significant, sustained declines early on. This visual emphasis makes their academic challenges readily apparent and highlights the crucial timing for interventions.
*   **RECOMMEND:** Utilize this refined visual evidence to enhance existing early warning systems by integrating smoothed assessment scores. If a student's smoothed score trend consistently matches or falls below the 'Withdrawn' trend line at critical junctures, this should trigger immediate and intensive interventions, such as personalized academic success plans, one-on-one advising with a dedicated student support specialist, or direct outreach from course coordinators.
